# Create Analytical Dataset

```python
df = orders.merge(
    customers,
    on="customerid",
    how="left"
)
```

```python
df = df.merge(
    order_details,
    on="orderid",
    how="inner"
)
```

```python
df = df.merge(
    products,
    on="productid",
    how="left"
)
```

```python
df.shape
```

```python
df.head()
```

```python
df.columns
```

```python
df["orderdetailid"].duplicated().sum()
```

```python
df["year"] = df["orderdate"].dt.year
```

```python
df["month"] = df["orderdate"].dt.month
```

```python
df["monthname"] = df["orderdate"].dt.month_name()
```

```python
df["yearmonth"] = df["orderdate"].dt.to_period("M").astype(str)
```

```python
df["profitmarginpct"] = (
    df["profit"] /
    df["salesamount"]
) * 100
```
